```python
import socket
import time

# ============================================================
# Agilent 34980A - Socket Communication
# Read DC voltage from a list of channels
# ============================================================

IP_ADDRESS = "192.168.1.75"
PORT = 5025

# Channels to measure
channels = [1031]

# Create socket
sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
sock.settimeout(5)

try:
    # --------------------------------------------------------
    # Connect to instrument
    # --------------------------------------------------------
    print(f"Connecting to {IP_ADDRESS}:{PORT}...")
    sock.connect((IP_ADDRESS, PORT))
    print("Connected")

    # --------------------------------------------------------
    # Identify instrument
    # --------------------------------------------------------
    sock.sendall(b"*IDN?\n")
    response = sock.recv(4096).decode().strip()

    print("Instrument:", response)

    print("\nMeasurements:")
    print("-------------------------")

    # --------------------------------------------------------
    # Scan through each channel
    # --------------------------------------------------------
    for channel in channels:

        print(f"Measuring Channel {channel}...")

        # Configure DC voltage measurement
        command = f"CONF:VOLT:DC (@{channel})\n"
        sock.sendall(command.encode())

        # Start measurement
        sock.sendall(b"INIT\n")

        # Allow measurement to complete
        time.sleep(0.5)

        # Fetch result
        sock.sendall(b"FETC?\n")

        response = sock.recv(4096).decode().strip()

        # Convert result
        try:
            voltage = float(response)
            print(f"Channel {channel}: {voltage:.6f} V")
        except ValueError:
            print(f"Channel {channel}: {response}")

finally:
    # --------------------------------------------------------
    # Close connection
    # --------------------------------------------------------
    sock.close()
    print("\nConnection closed")
```
