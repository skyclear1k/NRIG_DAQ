```python
import socket

# Configure instrument connection
IP_ADDRESS = "192.168.1.75"  # Change to your 34980A IP
PORT = 5025                   # Standard SCPI raw socket port

try:
    # Open TCP socket connection
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        s.settimeout(2.0)
        s.connect((IP_ADDRESS, PORT))
        
        # Send command (SCPI commands over raw sockets must end with \n)
        s.sendall(b"SYSTem:BEEPer:STATe OFF\n")
        print("Beeper disabled successfully.")
        
except socket.timeout:
    print("Connection timed out. Check the IP address and network layout.")
except Exception as e:
    print(f"An error occurred: {e}")
```
