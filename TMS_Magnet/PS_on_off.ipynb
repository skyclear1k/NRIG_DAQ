```python
import socket

HOST = "192.168.1.70"   # Power supply IP
PORT = 50505

VOLTAGE_LIMIT = 1       # V — maximum allowed voltage
CURRENT = 137            # A — desired current

with socket.create_connection((HOST, PORT), timeout=5) as s:

    # 1. Set maximum voltage
    s.sendall(f"VOLT {VOLTAGE_LIMIT}\n".encode())

    # 2. Assign desired current
    s.sendall(f"CURR {CURRENT}\n".encode())

    # 3. Enable output
    s.sendall(b"OUTP ON\n")

    print(
        f"Power supply enabled: "
        f"{CURRENT} A current, {VOLTAGE_LIMIT} V voltage limit"
    )
```
