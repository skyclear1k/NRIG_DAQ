```python
import socket

HOST = '192.168.1.70'  # instrument IP
PORT = 50505              # common SCPI raw socket port

sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
sock.settimeout(5)
sock.connect((HOST, PORT))

def query(cmd):
    sock.sendall((cmd + '\n').encode())
    return sock.recv(4096).decode().strip()

def write(cmd):
    sock.sendall((cmd + '\n').encode())

print(query('*IDN?'))       # identify the instrument
write('*RST')                # reset
write('VOLT 0.0')            # example: set voltage
print(query('VOLT?'))        # query it back
print(query('SOUR:CURR?'))    # legge il setpoint di corrente impostato
print(query('MEAS:CURR?'))    # legge la corrente realmente erogata (misura)

sock.close()
```
