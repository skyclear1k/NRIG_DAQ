```python
"""
Example showing a loop used to read continuosly from a module until script is interrupted or stopped.
"""
from maq20 import MAQ20
import time

module_to_use = "DIOL"  # change this to a module in your system. ("VSN", "VDN", etc)
delay_s = 0.1  # amount in seconds the script waits before reading another sample.


maq20 = MAQ20(ip_address="192.168.1.40", port=502)

a_module = maq20.find(module_to_use)

if a_module is None:  # Check if module was found.
    raise TypeError("Module not found.")

# print(a_module[0])
print(a_module[:])  # same as: a_module.read_data(0, number_of_channels=a_module.get_number_of_channels())
time.sleep(delay_s)
```

```python
"""
Read values continuously from MAQ20 module with registration number 6
"""

from maq20 import MAQ20
import time

IP_ADDRESS = "192.168.1.30"
PORT = 502
REGISTRATION_NUMBER = 1
DELAY_S = 0.1  # time between reads

maq20 = MAQ20(ip_address=IP_ADDRESS, port=PORT)

# Get module by registration number
module = maq20[REGISTRATION_NUMBER]

print("Connected to module:", module.read_name())
print("Number of channels:", module.get_number_of_channels())


# Read all channel values
values = module[:]

print(len(values))
print(values[0:7])
```
