```python
from pymodbus.client import ModbusTcpClient

IP = "192.168.1.40"
PORT = 502
R = 6   # registration number, only if 6 is truly the DIOL registration number

addr = 2000 * R + 1000
ch = 10 # channle 10 is wher I write the binary format [0,0,0,0,0] as decimal

client = ModbusTcpClient(IP, port=PORT)
client.connect()

#################################
# WRITE                          #
#################################
value = 4
# OPEN valve: close the switch on DO0
result = client.write_register(addr+ch, value, slave=1)   # adjust slave if needed

client.close()

#################################
# READ                          #
#################################
result = client.read_holding_registers(addr, 5, slave=1) # the second position is the number of channels to read

print(result.registers)

# if result.isError():
#     print("Read error")
# else:
#     value = result.registers[0]
#     print("DO0 state:", value)
```

```python
from pymodbus.client import ModbusTcpClient

IP = "192.168.1.40"
PORT = 502
R = 6   # registration number, only if 6 is truly the DIOL registration number

addr_do0 = 2000 * R + 1000

client = ModbusTcpClient(IP, port=PORT)
client.connect()

#################################
# WRITE                          #
#################################
# OPEN valve: close the switch on DO0
result = client.write_register(addr_do0, 1, slave=1)   # adjust slave if needed

# CLOSE valve: open the switch on DO0
# result = client.write_register(addr_do0, 1, slave=1)

client.close()

#################################
# READ                          #
#################################
result = client.read_holding_registers(13000, 1, slave=1)

if result.isError():
    print("Read error")
else:
    value = result.registers[0]
    print("DO0 state:", value)
```
