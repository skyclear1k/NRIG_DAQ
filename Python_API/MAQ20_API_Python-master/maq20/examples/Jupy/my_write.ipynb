```python
from pymodbus.client import ModbusTcpClient

MAQ20_IP = "192.168.1.30"
PORT = 502
SLAVE_ID = 1

REG_1BASED = 2100
VALUE = 1
ADDRESS = REG_1BASED - 1   # pymodbus = 0-based

client = ModbusTcpClient(MAQ20_IP, port=PORT, timeout=1.0)

if not client.connect():
    raise RuntimeError("Failed to connect to MAQ20")

result = client.write_register(
    address=ADDRESS,
    value=VALUE,
    slave=SLAVE_ID
)

client.close()

if result.isError():
    raise RuntimeError(f"Modbus write error: {result}")

print(f"Wrote {VALUE} to register {REG_1BASED}")
```
