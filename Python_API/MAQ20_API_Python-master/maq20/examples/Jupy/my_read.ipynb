```python
from pymodbus.client import ModbusTcpClient

MAQ20_IP = "192.168.1.30"   # <-- your MAQ20 IP
PORT = 502
UNIT_ID = 1
# REGISTER = 2100            # as requested
REGISTER = 3000            # as requested

client = ModbusTcpClient(MAQ20_IP, port=PORT)

if not client.connect():
    raise RuntimeError("Failed to connect to MAQ20")

result = client.read_holding_registers(
    address=REGISTER,
    count=8,
    slave=UNIT_ID
)

client.close()

if result.isError():
    raise RuntimeError(f"Modbus error: {result}")

value = result.registers[0:7]
print(f"Register {REGISTER} = {value}")
```
